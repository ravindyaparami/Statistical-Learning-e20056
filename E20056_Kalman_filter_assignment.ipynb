# Q. Analytical Derivation

Consider the filter model
\begin{align*}
x^{-}_k &= A_{k-1}\,x^{+}_{k-1} + G_{k-1}\,w_{k-1}, \\
y^{-}_k &= H_k\,x^{-}_k + z_k,
\end{align*}
where
$$x^{+}_{k-1} \sim \mathscr{N}(m_{k-1}, P_{k-1}),$$

Answer the followings:

1. Show that
$$x_k^- \sim \mathscr{N}(m_k^-, P_k^-),$$
where
$$m_k^- = A_{k-1}m_{k-1},$$
and
$$
P_k^- = A_{k-1}P_{k-1}A_{k-1}^T + G_{k-1}\Sigma_p G_{k-1}^T.
$$

2. Show that
$$y_k^- \sim \mathscr{N}(H_k m_k^-, H_k P_k^- H_k^T + \Sigma_m).$$

3. Show that
$$\begin{bmatrix} x_k^- \\ y^{-}_k \end{bmatrix} \sim \mathscr{N}\left(
\begin{bmatrix} m_k^- \\ H_k m_k^- \end{bmatrix},
\begin{bmatrix} P_k^- & P_k^- H_k^T \\ H_k P_k^- & H_k P_k^- H_k^T + \Sigma_m \end{bmatrix}
\right).$$

4. Show that
$$x^{+}_k\triangleq (x^{-}_k \mid y^{-}_k = y^{\mathrm{obs}}_{k}) \sim \mathscr{N}(m_k, P_k),$$
where the updated mean $m_k$ and updated covariance $P_k$ are given by:
\begin{align*}
K_k &\triangleq P_k^- H_k^T (H_k P_k^- H_k^T + \Sigma_m)^{-1}, \\
m_k &= m_k^- + K_k (y^{\mathrm{obs}}_{k} - H_k m_k^-), \\
P_k &= (I - K_k H_k) P_k^-.
\end{align*}

5. Find
$\mathbb{E}[x^{-}_k \mid y^{-}_k = y^{\mathrm{obs}}_{k}]$  and
$\text{Var}(x^{-}_k \mid y^{-}_k = y^{\mathrm{obs}}_{k})$.

# Answer

Consider the filter model

$$
x_k^- = A_{k-1}x_{k-1}^+ + G_{k-1}w_{k-1}
$$

$$
y_k^- = H_kx_k^- + z_k
$$

where

$$
x_{k-1}^+ \sim \mathcal{N}(m_{k-1}, P_{k-1})
$$

Assume that

$$
w_{k-1} \sim \mathcal{N}(0, \Sigma_p)
$$

and

$$
z_k \sim \mathcal{N}(0, \Sigma_m)
$$

Also assume that (x_{k-1}^+), (w_{k-1}), and (z_k) are mutually independent.

---

## 1. Distribution of (x_k^-)

The prediction model is

$$
x_k^- = A_{k-1}x_{k-1}^+ + G_{k-1}w_{k-1}
$$

Since (x_{k-1}^+) and (w_{k-1}) are Gaussian random variables, and (x_k^-) is a linear combination of them, (x_k^-) is also Gaussian.

Therefore,

$$
x_k^- \sim \mathcal{N}(m_k^-, P_k^-)
$$

### Mean of (x_k^-)

$$
m_k^- = \mathbb{E}[x_k^-]
$$

$$
m_k^- = \mathbb{E}[A_{k-1}x_{k-1}^+ + G_{k-1}w_{k-1}]
$$

$$
m_k^- = A_{k-1}\mathbb{E}[x_{k-1}^+] + G_{k-1}\mathbb{E}[w_{k-1}]
$$

Since

$$
\mathbb{E}[x_{k-1}^+] = m_{k-1}
$$

and

$$
\mathbb{E}[w_{k-1}] = 0
$$

we get

$$
m_k^- = A_{k-1}m_{k-1}
$$

### Covariance of (x_k^-)

$$
P_k^- = \text{Var}(x_k^-)
$$

$$
P_k^- = \text{Var}(A_{k-1}x_{k-1}^+ + G_{k-1}w_{k-1})
$$

Since (x_{k-1}^+) and (w_{k-1}) are independent,

$$
P_k^- = A_{k-1}P_{k-1}A_{k-1}^T + G_{k-1}\Sigma_pG_{k-1}^T
$$

Therefore,

$$
x_k^- \sim \mathcal{N}(m_k^-, P_k^-)
$$

where

$$
m_k^- = A_{k-1}m_{k-1}
$$

and

$$
P_k^- = A_{k-1}P_{k-1}A_{k-1}^T + G_{k-1}\Sigma_pG_{k-1}^T
$$

---

## 2. Distribution of (y_k^-)

The measurement model is

$$
y_k^- = H_kx_k^- + z_k
$$

Since (x_k^-) is Gaussian and (z_k) is Gaussian, (y_k^-) is also Gaussian.

Therefore,

$$
y_k^- \sim \mathcal{N}(H_km_k^-, H_kP_k^-H_k^T + \Sigma_m)
$$

### Mean of (y_k^-)

$$
\mathbb{E}[y_k^-] = \mathbb{E}[H_kx_k^- + z_k]
$$

$$
\mathbb{E}[y_k^-] = H_k\mathbb{E}[x_k^-] + \mathbb{E}[z_k]
$$

Since

$$
\mathbb{E}[x_k^-] = m_k^-
$$

and

$$
\mathbb{E}[z_k] = 0
$$

we get

$$
\mathbb{E}[y_k^-] = H_km_k^-
$$

### Covariance of (y_k^-)

$$
\text{Var}(y_k^-) = \text{Var}(H_kx_k^- + z_k)
$$

Since (x_k^-) and (z_k) are independent,

$$
\text{Var}(y_k^-) = H_kP_k^-H_k^T + \Sigma_m
$$

Therefore,

$$
y_k^- \sim \mathcal{N}(H_km_k^-, H_kP_k^-H_k^T + \Sigma_m)
$$

---

## 3. Joint Distribution of (x_k^-) and (y_k^-)

Since (y_k^-) is a linear function of (x_k^-) plus Gaussian noise, the vector

$$
\begin{bmatrix}
x_k^- \
y_k^-
\end{bmatrix}
$$

is jointly Gaussian.

The mean vector is

$$
\mathbb{E}
\begin{bmatrix}
x_k^- \
y_k^-
\end{bmatrix}
=============

\begin{bmatrix}
m_k^- \
H_km_k^-
\end{bmatrix}
$$

Now we find the covariance blocks.

First,

$$
\text{Var}(x_k^-) = P_k^-
$$

Next,

$$
\text{Cov}(x_k^-, y_k^-)
========================

\text{Cov}(x_k^-, H_kx_k^- + z_k)
$$

Since (z_k) is independent of (x_k^-),

$$
\text{Cov}(x_k^-, z_k) = 0
$$

Therefore,

$$
\text{Cov}(x_k^-, y_k^-) = P_k^-H_k^T
$$

Similarly,

$$
\text{Cov}(y_k^-, x_k^-) = H_kP_k^-
$$

Finally,

$$
\text{Var}(y_k^-) = H_kP_k^-H_k^T + \Sigma_m
$$

Therefore,

$$
\begin{bmatrix}
x_k^- \
y_k^-
\end{bmatrix}
\sim
\mathcal{N}
\left(
\begin{bmatrix}
m_k^- \
H_km_k^-
\end{bmatrix},
\begin{bmatrix}
P_k^- & P_k^-H_k^T \
H_kP_k^- & H_kP_k^-H_k^T + \Sigma_m
\end{bmatrix}
\right)
$$

---

## 4. Conditional Distribution of (x_k^-) Given (y_k^- = y_k^{obs})

For a jointly Gaussian random vector,

$$
\begin{bmatrix}
x \
y
\end{bmatrix}
\sim
\mathcal{N}
\left(
\begin{bmatrix}
m_x \
m_y
\end{bmatrix},
\begin{bmatrix}
\Sigma_{xx} & \Sigma_{xy} \
\Sigma_{yx} & \Sigma_{yy}
\end{bmatrix}
\right)
$$

the conditional distribution of (x) given (y = y_{obs}) is

$$
x|y=y_{obs}
\sim
\mathcal{N}
\left(
m_x + \Sigma_{xy}\Sigma_{yy}^{-1}(y_{obs}-m_y),
\Sigma_{xx} - \Sigma_{xy}\Sigma_{yy}^{-1}\Sigma_{yx}
\right)
$$

In our case,

$$
m_x = m_k^-
$$

$$
m_y = H_km_k^-
$$

$$
\Sigma_{xx} = P_k^-
$$

$$
\Sigma_{xy} = P_k^-H_k^T
$$

$$
\Sigma_{yx} = H_kP_k^-
$$

and

$$
\Sigma_{yy} = H_kP_k^-H_k^T + \Sigma_m
$$

Therefore,

$$
x_k^- | y_k^- = y_k^{obs}
\sim
\mathcal{N}(m_k, P_k)
$$

where

$$
m_k =
m_k^- +
P_k^-H_k^T
(H_kP_k^-H_k^T + \Sigma_m)^{-1}
(y_k^{obs} - H_km_k^-)
$$

Define the Kalman gain as

$$
K_k =
P_k^-H_k^T
(H_kP_k^-H_k^T + \Sigma_m)^{-1}
$$

Then the updated mean becomes

$$
m_k = m_k^- + K_k(y_k^{obs} - H_km_k^-)
$$

The updated covariance is

$$
P_k =
P_k^-
-----

P_k^-H_k^T
(H_kP_k^-H_k^T + \Sigma_m)^{-1}
H_kP_k^-
$$

Using the Kalman gain,

$$
P_k = P_k^- - K_kH_kP_k^-
$$

Therefore,

$$
P_k = (I - K_kH_k)P_k^-
$$

---

## 5. Final Results

The conditional expectation is

$$
\mathbb{E}[x_k^- | y_k^- = y_k^{obs}]
=====================================

m_k^- + K_k(y_k^{obs} - H_km_k^-)
$$

where

$$
K_k =
P_k^-H_k^T
(H_kP_k^-H_k^T + \Sigma_m)^{-1}
$$

Therefore,

$$
\mathbb{E}[x_k^- | y_k^- = y_k^{obs}] = m_k
$$

The conditional variance is

$$
\text{Var}(x_k^- | y_k^- = y_k^{obs})
=====================================

## P_k^-

P_k^-H_k^T
(H_kP_k^-H_k^T + \Sigma_m)^{-1}
H_kP_k^-
$$

Equivalently,

$$
\text{Var}(x_k^- | y_k^- = y_k^{obs})
=====================================

(I - K_kH_k)P_k^-
$$

Therefore,

$$
\text{Var}(x_k^- | y_k^- = y_k^{obs}) = P_k
$$


# Q. 1-D Example

Consider the scalar linear-Gaussian filter model
\begin{aligned}
x^-_k &= a\,x^+_{k-1} + w_{k-1},\qquad w_{k-1}\sim\mathscr N(0,q),\\
y^-_k &= h\,x^-_k + z_k,\qquad\;\;\;\; z_k\sim\mathscr N(0,r),
\end{aligned}
with prior $x_0\sim \mathscr{N}(m_0,P_0)$. Define $Y_{k}=\{y^{\mathrm{obs}}_1,\dots,y^{\mathrm{obs}}_{k}\}$.

Answer the following:

1. Show that
\begin{aligned}
m_k^- &= a\,m_{k-1},\\
P_k^- &= a^2 P_{k-1} + q.
\end{aligned}

2. Show that
\begin{aligned}
m_k &= m_k^- + K_k\,v_k
= m_k^- + \frac{P_k^- h}{S_k}\bigl(y^{\mathrm{obs}}_k - h\,m_k^-\bigr),\\
P_k &= (1 - K_k h)\,P_k^-
= \Bigl(1 - \frac{P_k^- h^2}{S_k}\Bigr) P_k^- .
\end{aligned}

3. Show that the predictive measurement distribution (before seeing $y_k$):
\begin{align*}
p(y^-_k | Y_{k-1})=\mathscr N\bigl(h\,m_k^-,\; h^2 P_k^- + r\bigr).
\end{align*}

4. Show that the posterior-predictive measurement distribution (after filtering on $y_k$):
\begin{align*}
p(y^-_k\mid Y_k)=\mathscr N\bigl(h\,m_k,\; h^2 P_k + r\bigr).
\end{align*}

5. For suitable numerical values for $a,q,h,r$ animate the above to prior distribution and the posterior distribution.

# Answer

Consider the scalar linear-Gaussian filter model

$$
x_k^- = ax_{k-1}^+ + w_{k-1}
$$

where

$$
w_{k-1} \sim \mathcal{N}(0,q)
$$

and

$$
y_k^- = hx_k^- + z_k
$$

where

$$
z_k \sim \mathcal{N}(0,r)
$$

The initial prior is

$$
x_0 \sim \mathcal{N}(m_0,P_0)
$$

Let

$$
Y_k = {y_1^{obs}, y_2^{obs}, \dots, y_k^{obs}}
$$

---

## 1. Prediction Step

Assume that after the previous measurement update,

$$
x_{k-1}^+ | Y_{k-1} \sim \mathcal{N}(m_{k-1},P_{k-1})
$$

The process model is

$$
x_k^- = ax_{k-1}^+ + w_{k-1}
$$

Since (x_{k-1}^+) and (w_{k-1}) are Gaussian, (x_k^-) is also Gaussian.

Therefore,

$$
x_k^- | Y_{k-1} \sim \mathcal{N}(m_k^-,P_k^-)
$$

### Predicted Mean

$$
m_k^- = \mathbb{E}[x_k^- | Y_{k-1}]
$$

$$
m_k^- = \mathbb{E}[ax_{k-1}^+ + w_{k-1}]
$$

$$
m_k^- = a\mathbb{E}[x_{k-1}^+] + \mathbb{E}[w_{k-1}]
$$

Since

$$
\mathbb{E}[x_{k-1}^+] = m_{k-1}
$$

and

$$
\mathbb{E}[w_{k-1}] = 0
$$

we get

$$
m_k^- = am_{k-1}
$$

### Predicted Covariance

$$
P_k^- = \text{Var}(x_k^- | Y_{k-1})
$$

$$
P_k^- = \text{Var}(ax_{k-1}^+ + w_{k-1})
$$

Since (x_{k-1}^+) and (w_{k-1}) are independent,

$$
P_k^- = a^2P_{k-1} + q
$$

Therefore,

$$
m_k^- = am_{k-1}
$$

and

$$
P_k^- = a^2P_{k-1} + q
$$

---

## 2. Measurement Update Step

The measurement model is

$$
y_k^- = hx_k^- + z_k
$$

Before observing (y_k^{obs}), the predicted measurement mean is

$$
hm_k^-
$$

and the predicted measurement variance is

$$
S_k = h^2P_k^- + r
$$

Here,

$$
S_k
$$

is called the innovation variance.

The innovation is

$$
v_k = y_k^{obs} - hm_k^-
$$

The scalar Kalman gain is

$$
K_k = \frac{P_k^-h}{S_k}
$$

Since

$$
S_k = h^2P_k^- + r
$$

we can also write

$$
K_k = \frac{P_k^-h}{h^2P_k^- + r}
$$

The updated mean is

$$
m_k = m_k^- + K_kv_k
$$

Substituting the innovation,

$$
m_k = m_k^- + K_k(y_k^{obs} - hm_k^-)
$$

Substituting the Kalman gain,

$$
m_k = m_k^- + \frac{P_k^-h}{S_k}(y_k^{obs} - hm_k^-)
$$

The updated covariance is

$$
P_k = (1-K_kh)P_k^-
$$

Substituting

$$
K_k = \frac{P_k^-h}{S_k}
$$

we get

$$
P_k = \left(1-\frac{P_k^-h^2}{S_k}\right)P_k^-
$$

Therefore,

$$
m_k = m_k^- + K_kv_k
$$

where

$$
v_k = y_k^{obs} - hm_k^-
$$

and

$$
K_k = \frac{P_k^-h}{S_k}
$$

Also,

$$
P_k = (1-K_kh)P_k^-
$$

or equivalently,

$$
P_k = \left(1-\frac{P_k^-h^2}{S_k}\right)P_k^-
$$

---

## 3. Predictive Measurement Distribution Before Seeing (y_k)

Before observing (y_k^{obs}), we have

$$
x_k^- | Y_{k-1} \sim \mathcal{N}(m_k^-,P_k^-)
$$

The measurement model is

$$
y_k^- = hx_k^- + z_k
$$

where

$$
z_k \sim \mathcal{N}(0,r)
$$

Since (y_k^-) is a linear function of a Gaussian random variable plus Gaussian noise, (y_k^-) is Gaussian.

The mean is

$$
\mathbb{E}[y_k^- | Y_{k-1}] = hm_k^-
$$

The variance is

$$
\text{Var}(y_k^- | Y_{k-1}) = h^2P_k^- + r
$$

Therefore,

$$
p(y_k^- | Y_{k-1}) =
\mathcal{N}(hm_k^-, h^2P_k^- + r)
$$

---

## 4. Posterior-Predictive Measurement Distribution After Filtering on (y_k)

After filtering using (y_k^{obs}), we have

$$
x_k^+ | Y_k \sim \mathcal{N}(m_k,P_k)
$$

Using the measurement equation again,

$$
y_k^- = hx_k^+ + z_k
$$

where

$$
z_k \sim \mathcal{N}(0,r)
$$

The mean is

$$
\mathbb{E}[y_k^- | Y_k] = hm_k
$$

The variance is

$$
\text{Var}(y_k^- | Y_k) = h^2P_k + r
$$

Therefore,

$$
p(y_k^- | Y_k) =
\mathcal{N}(hm_k, h^2P_k + r)
$$

---

## 5. Summary of Scalar Kalman Filter Equations

### Prediction

$$
m_k^- = am_{k-1}
$$

$$
P_k^- = a^2P_{k-1} + q
$$

### Innovation

$$
v_k = y_k^{obs} - hm_k^-
$$

### Innovation Variance

$$
S_k = h^2P_k^- + r
$$

### Kalman Gain

$$
K_k = \frac{P_k^-h}{S_k}
$$

### Update

$$
m_k = m_k^- + K_kv_k
$$

$$
P_k = (1-K_kh)P_k^-
$$

or

$$
P_k = \left(1-\frac{P_k^-h^2}{S_k}\right)P_k^-
$$

### Predictive Measurement Distribution

$$
p(y_k^- | Y_{k-1}) =
\mathcal{N}(hm_k^-, h^2P_k^- + r)
$$

### Posterior-Predictive Measurement Distribution

$$
p(y_k^- | Y_k) =
\mathcal{N}(hm_k, h^2P_k + r)
$$


# Q. 2D-Position Estimation

## Part A

*Consider* a two-dimensional constant-velocity dynamical system. The hidden state at time step $k$ is

$$
x_k =
\begin{bmatrix}
p_x(k)\\
p_y(k)\\
v_x(k)\\
v_y(k)
\end{bmatrix},
$$

where $(p_x(k),p_y(k))$ denote the position components and $(v_x(k),v_y(k))$ denote the velocity components.

The measurement consists only of the two position components:

$$
y_k =
\begin{bmatrix}
p_x^{\mathrm{meas}}(k)\\
p_y^{\mathrm{meas}}(k)
\end{bmatrix}.
$$

The linear Gaussian state-space filter model is

$$
x^-_k = A x^+_{k-1} + G w_{k-1},
$$

$$
y^-_k = Hx^+_k + z_k,
$$

where

$$
w_{k-1} \sim \mathscr{N}(0,\Sigma_p),
\qquad
z_k \sim \mathscr{N}(0,\Sigma_m).
$$

The process noise sequence $w_k$, measurement noise sequence $z_k$, and the initial state are assumed mutually independent.

Show that:

$$
A =
\begin{bmatrix}
1 & 0 & \Delta t & 0\\
0 & 1 & 0 & \Delta t\\
0 & 0 & 1 & 0\\
0 & 0 & 0 & 1
\end{bmatrix},
$$



$$
H =
\begin{bmatrix}
1 & 0 & 0 & 0\\
0 & 1 & 0 & 0
\end{bmatrix},
$$

and

$$
G =
\begin{bmatrix}
\frac{1}{2}\Delta t^2 & 0\\
0 & \frac{1}{2}\Delta t^2\\
\Delta t & 0\\
0 & \Delta t
\end{bmatrix}.
$$

# Answer

The hidden state at time step (k) is

$$
x_k =
\begin{bmatrix}
p_x(k)\
p_y(k)\
v_x(k)\
v_y(k)
\end{bmatrix}
$$

where (p_x(k)) and (p_y(k)) are the position components, and (v_x(k)) and (v_y(k)) are the velocity components.

The measurement consists only of the position components:

$$
y_k =
\begin{bmatrix}
p_x^{\mathrm{meas}}(k)\
p_y^{\mathrm{meas}}(k)
\end{bmatrix}
$$

We consider the linear Gaussian state-space model

$$
x_k^- = Ax_{k-1}^+ + Gw_{k-1}
$$

and

$$
y_k^- = Hx_k^- + z_k
$$

where

$$
w_{k-1} \sim \mathcal{N}(0,\Sigma_p)
$$

and

$$
z_k \sim \mathcal{N}(0,\Sigma_m)
$$

Here, (w_{k-1}) represents the process noise and (z_k) represents the measurement noise.

---

## 1. Derivation of the State Transition Matrix (A)

For a constant-velocity model, the position at the next time step is obtained using

$$
p_x(k) = p_x(k-1) + \Delta t , v_x(k-1)
$$

$$
p_y(k) = p_y(k-1) + \Delta t , v_y(k-1)
$$

The velocity is assumed to remain constant from one time step to the next:

$$
v_x(k) = v_x(k-1)
$$

$$
v_y(k) = v_y(k-1)
$$

Therefore, the state equation can be written as

$$
\begin{bmatrix}
p_x(k)\
p_y(k)\
v_x(k)\
v_y(k)
\end{bmatrix}
=============

\begin{bmatrix}
1 & 0 & \Delta t & 0\
0 & 1 & 0 & \Delta t\
0 & 0 & 1 & 0\
0 & 0 & 0 & 1
\end{bmatrix}
\begin{bmatrix}
p_x(k-1)\
p_y(k-1)\
v_x(k-1)\
v_y(k-1)
\end{bmatrix}
$$

Hence, the state transition matrix is

$$
A =
\begin{bmatrix}
1 & 0 & \Delta t & 0\
0 & 1 & 0 & \Delta t\
0 & 0 & 1 & 0\
0 & 0 & 0 & 1
\end{bmatrix}
$$

---

## 2. Derivation of the Measurement Matrix (H)

The measurement contains only the position components (p_x(k)) and (p_y(k)). It does not directly measure the velocity components (v_x(k)) and (v_y(k)).

Therefore,

$$
y_k =
\begin{bmatrix}
p_x^{\mathrm{meas}}(k)\
p_y^{\mathrm{meas}}(k)
\end{bmatrix}
$$

can be modeled as

$$
y_k^- = Hx_k^- + z_k
$$

where

$$
x_k^- =
\begin{bmatrix}
p_x(k)\
p_y(k)\
v_x(k)\
v_y(k)
\end{bmatrix}
$$

Since the first measurement is (p_x(k)), it selects the first state variable:

$$
p_x^{\mathrm{meas}}(k) = p_x(k) + \text{noise}
$$

Since the second measurement is (p_y(k)), it selects the second state variable:

$$
p_y^{\mathrm{meas}}(k) = p_y(k) + \text{noise}
$$

Thus,

$$
\begin{bmatrix}
p_x^{\mathrm{meas}}(k)\
p_y^{\mathrm{meas}}(k)
\end{bmatrix}
=============

\begin{bmatrix}
1 & 0 & 0 & 0\
0 & 1 & 0 & 0
\end{bmatrix}
\begin{bmatrix}
p_x(k)\
p_y(k)\
v_x(k)\
v_y(k)
\end{bmatrix}
+
z_k
$$

Hence, the measurement matrix is

$$
H =
\begin{bmatrix}
1 & 0 & 0 & 0\
0 & 1 & 0 & 0
\end{bmatrix}
$$

---

## 3. Derivation of the Process Noise Matrix (G)

In a constant-velocity model, the process noise is commonly modeled as unknown acceleration affecting the motion.

Let

$$
w_{k-1} =
\begin{bmatrix}
a_x(k-1)\
a_y(k-1)
\end{bmatrix}
$$

where (a_x(k-1)) and (a_y(k-1)) are random acceleration disturbances in the (x)- and (y)-directions.

From the basic equations of motion,

$$
p_x(k) = p_x(k-1) + \Delta t,v_x(k-1) + \frac{1}{2}\Delta t^2 a_x(k-1)
$$

$$
p_y(k) = p_y(k-1) + \Delta t,v_y(k-1) + \frac{1}{2}\Delta t^2 a_y(k-1)
$$

and

$$
v_x(k) = v_x(k-1) + \Delta t,a_x(k-1)
$$

$$
v_y(k) = v_y(k-1) + \Delta t,a_y(k-1)
$$

Therefore, the process noise contribution to the state is

$$
Gw_{k-1}
========

\begin{bmatrix}
\frac{1}{2}\Delta t^2 a_x(k-1)\
\frac{1}{2}\Delta t^2 a_y(k-1)\
\Delta t,a_x(k-1)\
\Delta t,a_y(k-1)
\end{bmatrix}
$$

This can be written in matrix form as

$$
Gw_{k-1}
========

\begin{bmatrix}
\frac{1}{2}\Delta t^2 & 0\
0 & \frac{1}{2}\Delta t^2\
\Delta t & 0\
0 & \Delta t
\end{bmatrix}
\begin{bmatrix}
a_x(k-1)\
a_y(k-1)
\end{bmatrix}
$$

Hence,

$$
G =
\begin{bmatrix}
\frac{1}{2}\Delta t^2 & 0\
0 & \frac{1}{2}\Delta t^2\
\Delta t & 0\
0 & \Delta t
\end{bmatrix}
$$

---

## Final State-Space Model

Therefore, the two-dimensional constant-velocity system can be written as

$$
x_k^- = Ax_{k-1}^+ + Gw_{k-1}
$$

where

$$
A =
\begin{bmatrix}
1 & 0 & \Delta t & 0\
0 & 1 & 0 & \Delta t\
0 & 0 & 1 & 0\
0 & 0 & 0 & 1
\end{bmatrix}
$$

and

$$
G =
\begin{bmatrix}
\frac{1}{2}\Delta t^2 & 0\
0 & \frac{1}{2}\Delta t^2\
\Delta t & 0\
0 & \Delta t
\end{bmatrix}
$$

The measurement model is

$$
y_k^- = Hx_k^- + z_k
$$

where

$$
H =
\begin{bmatrix}
1 & 0 & 0 & 0\
0 & 1 & 0 & 0
\end{bmatrix}
$$

Thus, (A) describes how position changes due to velocity, (H) selects only the position components from the state vector, and (G) maps the acceleration noise into both position and velocity.


## Part B

Develop a python based method to filter a sequence of noisy GPS obtained position meaurements.

```python
import numpy as np
import matplotlib.pyplot as plt

# ---------------------------------------------------------
# 1. Convert latitude-longitude to local x-y coordinates
# ---------------------------------------------------------

def latlon_to_xy(lat, lon, lat0, lon0):
    """
    Convert latitude and longitude to local x-y coordinates in meters.
    This uses a simple local flat-earth approximation.
    Good for small areas such as vehicle/person tracking.
    """
    R = 6371000  # Earth radius in meters

    lat = np.radians(lat)
    lon = np.radians(lon)
    lat0 = np.radians(lat0)
    lon0 = np.radians(lon0)

    x = R * (lon - lon0) * np.cos(lat0)
    y = R * (lat - lat0)

    return x, y


def xy_to_latlon(x, y, lat0, lon0):
    """
    Convert local x-y coordinates back to latitude-longitude.
    """
    R = 6371000

    lat0_rad = np.radians(lat0)
    lon0_rad = np.radians(lon0)

    lat = y / R + lat0_rad
    lon = x / (R * np.cos(lat0_rad)) + lon0_rad

    return np.degrees(lat), np.degrees(lon)


# ---------------------------------------------------------
# 2. Kalman filter for GPS position measurements
# ---------------------------------------------------------

def kalman_filter_gps(gps_measurements, dt=1.0, gps_std=5.0, accel_std=1.0):
    """
    Filters noisy GPS measurements using a 2D constant-velocity Kalman filter.

    Parameters
    ----------
    gps_measurements : numpy array of shape (N, 2)
        GPS measurements as [latitude, longitude].

    dt : float
        Time step between measurements in seconds.

    gps_std : float
        Standard deviation of GPS measurement noise in meters.
        Example: 5 means GPS error is around 5 meters.

    accel_std : float
        Standard deviation of unknown acceleration noise.
        Larger value means the filter allows more sudden motion changes.

    Returns
    -------
    filtered_latlon : numpy array of shape (N, 2)
        Filtered GPS positions as [latitude, longitude].

    filtered_states : numpy array of shape (N, 4)
        Filtered states [px, py, vx, vy].
    """

    gps_measurements = np.asarray(gps_measurements)

    lat = gps_measurements[:, 0]
    lon = gps_measurements[:, 1]

    # Use first GPS point as local origin
    lat0 = lat[0]
    lon0 = lon[0]

    # Convert GPS lat-lon to local x-y coordinates
    x_meas, y_meas = latlon_to_xy(lat, lon, lat0, lon0)

    measurements_xy = np.column_stack((x_meas, y_meas))

    N = len(measurements_xy)

    # -----------------------------------------------------
    # State vector:
    # x = [px, py, vx, vy]^T
    # -----------------------------------------------------

    # State transition matrix
    A = np.array([
        [1, 0, dt, 0],
        [0, 1, 0, dt],
        [0, 0, 1, 0],
        [0, 0, 0, 1]
    ])

    # Measurement matrix
    # GPS measures only position: px and py
    H = np.array([
        [1, 0, 0, 0],
        [0, 1, 0, 0]
    ])

    # Process noise input matrix
    G = np.array([
        [0.5 * dt**2, 0],
        [0, 0.5 * dt**2],
        [dt, 0],
        [0, dt]
    ])

    # Process noise covariance
    Sigma_p = accel_std**2 * np.eye(2)

    Q = G @ Sigma_p @ G.T

    # Measurement noise covariance
    R = gps_std**2 * np.eye(2)

    # -----------------------------------------------------
    # Initial state
    # -----------------------------------------------------

    x = np.array([
        measurements_xy[0, 0],
        measurements_xy[0, 1],
        0,
        0
    ])

    # Initial covariance
    P = np.diag([gps_std**2, gps_std**2, 10, 10])

    filtered_states = []

    # -----------------------------------------------------
    # Kalman filter loop
    # -----------------------------------------------------

    for k in range(N):

        # Prediction step
        x_pred = A @ x
        P_pred = A @ P @ A.T + Q

        # Current GPS measurement
        y_obs = measurements_xy[k]

        # Innovation
        innovation = y_obs - H @ x_pred

        # Innovation covariance
        S = H @ P_pred @ H.T + R

        # Kalman gain
        K = P_pred @ H.T @ np.linalg.inv(S)

        # Update step
        x = x_pred + K @ innovation

        # Updated covariance
        P = (np.eye(4) - K @ H) @ P_pred

        filtered_states.append(x)

    filtered_states = np.array(filtered_states)

    # Convert filtered x-y back to latitude-longitude
    filtered_lat, filtered_lon = xy_to_latlon(
        filtered_states[:, 0],
        filtered_states[:, 1],
        lat0,
        lon0
    )

    filtered_latlon = np.column_stack((filtered_lat, filtered_lon))

    return filtered_latlon, filtered_states


# ---------------------------------------------------------
# 3. Example usage with noisy GPS data
# ---------------------------------------------------------

# Example: create artificial noisy GPS measurements
np.random.seed(10)

N = 60
dt = 1.0

# Starting location
lat_start = 7.2906
lon_start = 80.6337

# True motion in meters
true_x = np.linspace(0, 120, N)
true_y = np.linspace(0, 60, N)

# Convert true local x-y to latitude-longitude
true_lat, true_lon = xy_to_latlon(true_x, true_y, lat_start, lon_start)

# Add GPS noise
gps_noise_std_m = 6.0

noise_x = np.random.normal(0, gps_noise_std_m, N)
noise_y = np.random.normal(0, gps_noise_std_m, N)

noisy_lat, noisy_lon = xy_to_latlon(
    true_x + noise_x,
    true_y + noise_y,
    lat_start,
    lon_start
)

gps_measurements = np.column_stack((noisy_lat, noisy_lon))

# Apply Kalman filter
filtered_latlon, filtered_states = kalman_filter_gps(
    gps_measurements,
    dt=dt,
    gps_std=6.0,
    accel_std=1.0
)

# ---------------------------------------------------------
# 4. Plot results in local x-y coordinates
# ---------------------------------------------------------

# Convert results to local meters for plotting
noisy_x, noisy_y = latlon_to_xy(
    gps_measurements[:, 0],
    gps_measurements[:, 1],
    lat_start,
    lon_start
)

filtered_x, filtered_y = latlon_to_xy(
    filtered_latlon[:, 0],
    filtered_latlon[:, 1],
    lat_start,
    lon_start
)

plt.figure(figsize=(8, 6))
plt.plot(true_x, true_y, label="True path", linewidth=3)
plt.scatter(noisy_x, noisy_y, label="Noisy GPS measurements", s=25)
plt.plot(filtered_x, filtered_y, label="Kalman filtered path", linewidth=3)

plt.xlabel("x position / meters")
plt.ylabel("y position / meters")
plt.title("Filtering Noisy GPS Measurements Using a Kalman Filter")
plt.grid(True)
plt.legend()
plt.axis("equal")
plt.show()
```
