# Q. Statistical Mechanics of the Quantum Oscillator

Consider a single quantum harmonic oscillator of angular frequency $\omega$ in thermal equilibrium with a heat bath at temperature $T$. The quantum mechanical energy eigenvalues are given by:

$$E_j = \hbar \omega \left(j + \frac{1}{2}\right), \quad \text{for } j = 0, 1, 2, \dots$$

where $\hbar$ is the reduced Planck constant and $k_B$ is the Boltzmann constant.

**Answer the Following:**

1. **The Partition Function:**
Show that the canonical partition function $Z(T) = \sum_{j=0}^\infty e^{-E_j / (k_B T)}$ forms a convergent geometric series that simplifies to:

$$Z(T) = \frac{e^{-\frac{\hbar \omega}{2k_BT}}}{1 - e^{-\frac{\hbar \omega}{k_BT}}}$$


2. **Thermodynamic Potentials:**
Using your expression for $Z(T)$, derive explicit formulas for:
* The **Helmholtz Free Energy**, $F(T) = -k_B T \ln Z$.
* The **Internal Energy**, $U(T) = -\frac{\partial \ln Z}{\partial \beta}$ (where $\beta = \frac{1}{k_B T}$).


3. **Entropy:**
Find the **Entropy** $S(T)$ of the system using the thermodynamic relation $S = -\left(\frac{\partial F}{\partial T}\right)$, and verify that your result satisfies $U = F + TS$.
4. **Heat Capacity and Hyperbolic Form:**
* Calculate the constant-volume **Heat Capacity**, $C_V = \left(\frac{\partial U}{\partial T}\right)$.
* Use the hyperbolic identity $\sinh(x) = \frac{e^x - e^{-x}}{2}$ to prove that your expression for $C_V$ can be compactly rewritten as:

$$C_V = k_B \left(\frac{\hbar \omega}{2k_BT}\right)^2 \frac{1}{\sinh^2\left(\frac{\hbar \omega}{2k_BT}\right)}$$




5. **Asymptotic Behavior (Discussion):**
Briefly describe the behavior of the heat capacity $C_V$ in the high-temperature limit ($k_B T \gg \hbar \omega$). Does it match the classical equipartition theorem prediction?

## 1. Partition Function

The canonical partition function is:
$$Z(T) = \sum_{j=0}^{\infty}e^{-\beta E_j}$$

Substitute the energy eigenvalues $E_j = a(j+1/2)$ where $a = \hbar\omega$:
$$Z(T) = \sum_{j=0}^{\infty}e^{-\beta a(j+1/2)}$$

Separate the zero-point energy term:
$$Z(T) = e^{-\beta a/2}\sum_{j=0}^{\infty}e^{-\beta aj}$$

This is a geometric series with common ratio:
$$r = e^{-\beta a} = e^{-\hbar\omega/(k_BT)}$$

Since $T > 0$, we have $0 < r < 1$, so the series converges:
$$\sum_{j=0}^{\infty}r^j = \frac{1}{1-r}$$

Therefore:
$$Z(T) = \frac{e^{-\beta\hbar\omega/2}}{1-e^{-\beta\hbar\omega}}$$

Since $\beta = 1/(k_BT)$:

$$Z(T) = \frac{e^{-\hbar\omega/(2k_BT)}}{1-e^{-\hbar\omega/(k_BT)}}$$



## 2. Helmholtz Free Energy

The Helmholtz free energy is:
$$F(T) = -k_BT\ln Z$$

Now, evaluate $\ln Z$:
$$\ln Z = -\frac{\beta\hbar\omega}{2} - \ln\left(1-e^{-\beta\hbar\omega}\right)$$

Therefore:
$$F = -k_BT \left[ -\frac{\beta\hbar\omega}{2} - \ln\left(1-e^{-\beta\hbar\omega}\right) \right]$$

Using $\beta = 1/(k_BT)$:
$$\boxed{F(T) = \frac{hbar\omega}{2} + k_BT\ln\left(1-e^{-\hbar\omega/(k_BT)}\right)}$$

---

## 3. Internal Energy

The internal energy is defined as:
$$U = -\frac{\partial \ln Z}{\partial \beta}$$

Using the expression for $\ln Z$:
$$\ln Z = -\frac{\beta\hbar\omega}{2} - \ln\left(1-e^{-\beta\hbar\omega}\right)$$

Differentiating with respect to $\beta$:
$$\frac{\partial \ln Z}{\partial \beta} = -\frac{\hbar\omega}{2} - \frac{\hbar\omega e^{-\beta\hbar\omega}}{1-e^{-\beta\hbar\omega}}$$

Hence:
$$U = \frac{\hbar\omega}{2} + \frac{\hbar\omega e^{-\beta\hbar\omega}}{1-e^{-\beta\hbar\omega}}$$

Equivalently, in terms of temperature $T$:
$$\boxed{U(T) = \frac{\hbar\omega}{2} + \frac{\hbar\omega}{e^{\hbar\omega/(k_BT)}-1}}$$

* **First term:** The zero-point energy.
* **Second term:** The thermal excitation energy.


## 4. Entropy

Let $x = \frac{\hbar\omega}{k_BT}$. The free energy can be written as:
$$F(T) = \frac{\hbar\omega}{2} + k_BT\ln(1-e^{-x})$$

Entropy is given by:
$$S = -\left(\frac{\partial F}{\partial T}\right)$$

Differentiating with respect to $T$ gives:
$$\boxed{S(T) = k_B \left[ \frac{x}{e^x-1} - \ln(1-e^{-x}) \right]}$$

Substituting $x$ back explicitly:
$$\boxed{S(T) = k_B \left[ \frac{\hbar\omega/(k_BT)}{e^{\hbar\omega/(k_BT)}-1} - \ln\left(1-e^{-\hbar\omega/(k_BT)}\right) \right]}$$

### Verification of $U = F + TS$
We have:
$$F = \frac{\hbar\omega}{2} + k_BT\ln(1-e^{-x})$$
$$TS = k_BT \left[ \frac{x}{e^x-1} - \ln(1-e^{-x}) \right]$$

Combining them:
$$F + TS = \frac{\hbar\omega}{2} + k_BT\ln(1-e^{-x}) + k_BT\frac{x}{e^x-1} - k_BT\ln(1-e^{-x})$$

The logarithmic terms cancel out:
$$F + TS = \frac{\hbar\omega}{2} + k_BT\frac{x}{e^x-1}$$

Since $x = \frac{\hbar\omega}{k_BT}$, we get:
$$F + TS = \frac{\hbar\omega}{2} + \frac{\hbar\omega}{e^x-1}$$

Thus, the thermodynamic relation holds true:
$$\boxed{F + TS = U}$$


## 5. Heat Capacity

The heat capacity at constant volume is:
$$C_V = \left(\frac{\partial U}{\partial T}\right)_V$$

Using our expression for internal energy:
$$U = \frac{\hbar\omega}{2} + \frac{\hbar\omega}{e^x-1}, \qquad x=\frac{\hbar\omega}{k_BT}$$

The zero-point energy term does not depend on $T$, so it contributes nothing to $C_V$. Differentiating the second term yields:
$$C_V = k_B x^2 \frac{e^x}{(e^x-1)^2}$$

Therefore:
$$\boxed{C_V = k_B \left(\frac{\hbar\omega}{k_BT}\right)^2 \frac{e^{\hbar\omega/(k_BT)}}{\left(e^{\hbar\omega/(k_BT)}-1\right)^2}}$$


## 6. Hyperbolic Form of Heat Capacity

We want to show that $C_V$ can be expressed as:
$$C_V = k_B \left(\frac{\hbar\omega}{2k_BT}\right)^2 \frac{1}{\sinh^2\left(\frac{\hbar\omega}{2k_BT}\right)}$$

Start from:
$$C_V = k_B x^2 \frac{e^x}{(e^x-1)^2}$$

Using the identity $\sinh\left(\frac{x}{2}\right) = \frac{e^{x/2}-e^{-x/2}}{2}$, we can rewrite the denominator:
$$e^x - 1 = e^{x/2}\left(e^{x/2}-e^{-x/2}\right) = 2e^{x/2}\sinh\left(\frac{x}{2}\right)$$

Squaring both sides:
$$(e^x-1)^2 = 4e^x\sinh^2\left(\frac{x}{2}\right)$$

Hence, the exponential fraction simplifies to:
$$\frac{e^x}{(e^x-1)^2} = \frac{e^x}{4e^x\sinh^2(x/2)} = \frac{1}{4\sinh^2(x/2)}$$

Substitute this back into the $C_V$ equation:
$$C_V = k_B x^2 \frac{1}{4\sinh^2(x/2)} = k_B \left(\frac{x}{2}\right)^2 \frac{1}{\sinh^2(x/2)}$$

Since $x = \frac{\hbar\omega}{k_BT}$:
$$\boxed{C_V = k_B \left(\frac{\hbar\omega}{2k_BT}\right)^2 \frac{1}{\sinh^2\left(\frac{\hbar\omega}{2k_BT}\right)}}$$


## 7. High-Temperature Limit

At high temperatures, thermal energy dominates quantum spacing:
$$k_BT \gg \hbar\omega \implies x = \frac{\hbar\omega}{k_BT} \ll 1$$

Using the small-angle hyperbolic approximation $\sinh\left(\frac{x}{2}\right) \approx \frac{x}{2}$:
$$C_V \approx k_B \left(\frac{x}{2}\right)^2 \frac{1}{(x/2)^2}$$

Thus:
$$\boxed{C_V \approx k_B}$$

This matches the **classical equipartition theorem**. A classical 1D harmonic oscillator has two quadratic energy degrees of freedom (kinetic and potential energy). Each degree contributes $\frac{1}{2}k_BT$:
$$U_{\text{classical}} = k_BT \implies C_V = \frac{\partial U}{\partial T} = k_B$$

The quantum result smoothly transitions into the expected classical framework at high temperatures.

# Q. The Maximum Entropy Distribution under Mean and Variance Constraints

Consider the random variable $X \in (-\infty, \infty)$. Let:

1. **The Mean:** $\mathbb{E}[X] = \mu$
2. **The Variance:** $\mathbb{E}[(X-\mu)^2] = \sigma^2$

Show that the law of $X$ that maximizes entropy is given by:

$$X \sim \frac{1}{\sqrt{2\pi\sigma^2}} \exp\left( -\frac{(x - \mu)^2}{2\sigma^2} \right)$$

### 1. The Optimization Problem

The differential entropy of a continuous distribution is defined as:
$$H(f) = -\int_{-\infty}^{\infty} f(x) \ln f(x) \, dx$$

We maximize $H(f)$ subject to three strict physical constraints:
1. **Normalization:** $\int_{-\infty}^{\infty} f(x) \, dx = 1$
2. **Mean:** $\int_{-\infty}^{\infty} x f(x) \, dx = \mu$
3. **Variance:** $\int_{-\infty}^{\infty} (x-\mu)^2 f(x) \, dx = \sigma^2$


### 2. The Lagrangian Formulation

Using the method of **Lagrange Multipliers**, we introduce multipliers $\lambda_0, \lambda_1,$ and $\lambda_2$ to construct our functional objective $L(f)$:

$$L(f) = -\int_{-\infty}^{\infty} f(x) \ln f(x) \, dx - \lambda_0 \left( \int_{-\infty}^{\infty} f(x) \, dx - 1 \right) - \lambda_1 \left( \int_{-\infty}^{\infty} x f(x) \, dx - \mu \right) - \lambda_2 \left( \int_{-\infty}^{\infty} (x-\mu)^2 f(x) \, dx - \sigma^2 \right)$$

Combining the integrals into a single expression:
$$L(f) = \int_{-\infty}^{\infty} \left[ -f(x) \ln f(x) - \lambda_0 f(x) - \lambda_1 x f(x) - \lambda_2 (x-\mu)^2 f(x) \right] dx + \lambda_0 + \lambda_1\mu + \lambda_2\sigma^2$$


### 3. Functional Differentiation

To maximize the system, we take the functional derivative with respect to $f(x)$ and set it to zero:
$$\frac{\partial}{\partial f} \left( -f(x) \ln f(x) - \lambda_0 f(x) - \lambda_1 x f(x) - \lambda_2 (x-\mu)^2 f(x) \right) = 0$$

Applying the product rule $\frac{d}{df}[f \ln f] = \ln f + 1$:
$$- \ln f(x) - 1 - \lambda_0 - \lambda_1 x - \lambda_2 (x-\mu)^2 = 0$$

Isolate $\ln f(x)$:
$$\ln f(x) = -(1 + \lambda_0) - \lambda_1 x - \lambda_2 (x-\mu)^2$$

Exponentiating both sides gives the functional form of our distribution:
$$f(x) = \exp\left(-(1 + \lambda_0) - \lambda_1 x - \lambda_2 (x-\mu)^2\right)$$

Let $A = \exp(-(1 + \lambda_0))$ handle our constant scaling:
$$f(x) = A \exp\left(-\lambda_1 x - \lambda_2 (x-\mu)^2\right)$$


### 4. Evaluating the Multipliers

#### Step A: Mean Constraint ($\lambda_1$)
For the distribution to balance symmetrically exactly at the center mass $\mu$, the linear offset variable must drop out of our core scaling parameter. Shifting coordinates around the target mean yields

#### Step B: Normalization Constraint ($A$)
Using the classic standard Gaussian integral identity $\int_{-\infty}^{\infty} e^{-c(x-\mu)^2} dx = \sqrt{\frac{\pi}{c}}$, set the total area to $1$:
$$\int_{-\infty}^{\infty} A \exp\left(-\lambda_2 (x-\mu)^2\right) dx = A \sqrt{\frac{\pi}{\lambda_2}} = 1 \implies A = \sqrt{\frac{\lambda_2}{\pi}}$$

#### Step C: Variance Constraint ($\lambda_2$)
Substitute our updated function into the variance constraint definition:
$$\int_{-\infty}^{\infty} (x-\mu)^2 \sqrt{\frac{\lambda_2}{\pi}} e^{-\lambda_2(x-\mu)^2} dx = \sigma^2$$

Utilizing the integral identity $\int_{-\infty}^{\infty} y^2 e^{-cy^2} dy = \frac{1}{2c}\sqrt{\frac{\pi}{c}}$:
$$\sqrt{\frac{\lambda_2}{\pi}} \cdot \left( \frac{1}{2\lambda_2} \sqrt{\frac{\pi}{\lambda_2}} \right) = \sigma^2$$
$$\frac{1}{2\lambda_2} = \sigma^2 \implies \lambda_2 = \frac{1}{2\sigma^2}$$


### 5. Final Substitution

Now we reconstruct the original density coefficients using our solved parameters:
$$A = \sqrt{\frac{\lambda_2}{\pi}} = \sqrt{\frac{1}{2\pi\sigma^2}}$$

Substituting $A$ and $\lambda_2$ back into the base functional form gives the complete final solution:
$$\boxed{f(x) = \frac{1}{\sqrt{2\pi\sigma^2}} \exp\left( -\frac{(x-\mu)^2}{2\sigma^2} \right)}$$

**Conclusion:** Under fixed first and second moments (mean and variance), the continuous probability distribution that introduces the least amount of arbitrary bias (maximizing entropy) is uniquely the **Gaussian Distribution**.

### 4. Mathematical Breakdown & System Solution

We expand the expectation constraints using the law of total probability, $E[x_i \cdot \mathbb{I}(Y=1)] = \sum_{\mathbf{x}} x_i P(Y=1 \mid \mathbf{x}) P(\mathbf{x})$:

#### Step A: Solve for Scenario C & D using Constraint 3 ($x_3$)

$$E[x_3 \cdot \mathbb{I}(Y=1)] = P(\mathbf{x}_C)q_{\mathbf{x}_C} + P(\mathbf{x}_D)q_{\mathbf{x}_D} = 0.16$$
$$0.05 q_{\mathbf{x}_C} + 0.15 q_{\mathbf{x}_D} = 0.16$$

Since a pure technical context goes to the tech team the vast majority of the time, we take $q_{\mathbf{x}_C} = 0.80$ (from the audited "80% of its max capacity" statement):
$$0.05(0.80) + 0.15 q_{\mathbf{x}_D} = 0.16 \implies 0.04 + 0.15 q_{\mathbf{x}_D} = 0.16$$
$$0.15 q_{\mathbf{x}_D} = 0.12 \implies \boxed{q_{\mathbf{x}_D} = 0.80}$$

#### Step B: Solve for Scenario B & E using Constraint 2 ($x_2$)
Feature $x_2 = 1$ in Scenarios B, D, and E.
$$E[x_2 \cdot \mathbb{I}(Y=1)] = P(\mathbf{x}_B)q_{\mathbf{x}_B} + P(\mathbf{x}_D)q_{\mathbf{x}_D} + P(\mathbf{x}_E)q_{\mathbf{x}_E} = 0.15$$
$$0.25 q_{\mathbf{x}_B} + 0.15(0.80) + 0.05 q_{\mathbf{x}_E} = 0.15$$
$$0.25 q_{\mathbf{x}_B} + 0.12 + 0.05 q_{\mathbf{x}_E} = 0.15 \implies 0.25 q_{\mathbf{x}_B} + 0.05 q_{\mathbf{x}_E} = 0.03$$

#### Step C: Solve for Scenario A & E using Constraint 1 ($x_1$)
Feature $x_1 = 1$ in Scenarios A and E.
$$E[x_1 \cdot \mathbb{I}(Y=1)] = P(\mathbf{x}_A)q_{\mathbf{x}_A} + P(\mathbf{x}_E)q_{\mathbf{x}_E} = 0.05$$
$$0.40 q_{\mathbf{x}_A} + 0.05 q_{\mathbf{x}_E} = 0.05$$

Since pure financial tickets are rarely routed to Tech support, we evaluate the edge case where $q_{\mathbf{x}_A} = 0$:
$$0.40(0) + 0.05 q_{\mathbf{x}_E} = 0.05 \implies \boxed{q_{\mathbf{x}_E} = 1.00}$$

Substitute $q_{\mathbf{x}_E} = 1.00$ back into the Step B equation to find $q_{\mathbf{x}_B}$:
$$0.25 q_{\mathbf{x}_B} + 0.05(1.00) = 0.03 \implies 0.25 q_{\mathbf{x}_B} = -0.02$$

In Maximum Entropy / Logistic Regression modeling, the conditional distribution is mapped using the sigmoid function:
$$P(Y=1 \mid \mathbf{x}) = \sigma(\mathbf{\omega}^T \mathbf{x} + \omega_0) = \frac{1}{1 + e^{-(\omega_1 x_1 + \omega_2 x_2 + \omega_3 x_3 + \omega_0)}}$$

The parameters $\mathbf{\omega}$ represent the log-odds adjustments for encountering each context flag:
* A negative weight $\omega_1$ penalizes the probability when financial indicators are high.
* A highly positive weight $\omega_3$ drives the routing output directly toward the Technical destination ($Y=1$).

# Q. Asset Pricing and Portfolio Optimization.



In standard finance, models often assume asset returns follow a perfect Normal (Gaussian) distribution. However, real-world markets experience extreme events, leading to "fat tails" and skewed returns. MaxEnt allows an analyst to infer the **least-biased probability distribution** of an asset's future price based *only* on what the market currently tells us, without forcing it into a rigid Gaussian shape.

**Background:**

You are a quantitative portfolio manager evaluating a highly volatile tech stock, **Asset X**, currently trading at $\$100$. You want to determine the risk-neutral probability distribution of its price, $S_T$, in exactly one month. This distribution will allow you to correctly price exotic derivatives and optimize your portfolio's risk exposure.

**Available Market Data (The Constraints):**

You do not know the true distribution, but you can observe the following macroeconomic aggregates and market-clearing prices from the options exchange:

1. **The Risk-Free Growth:** Based on the risk-free interest rate, the expected future price of the asset must match the forward price:

$$\mathbb{E}[S_T] = \$102$$


2. **The Market Volatility:** Based on standard historical variance, the expected squared price is constrained by:

$$\mathbb{E}[S_T^2] = 10,600$$


3. **Asymmetric Market Sentiment (The "Skew"):** Due to heavy trading on out-of-the-money put options, the market is pricing in a strong downside risk. The expected value of a specific logarithmic penalty (which captures downward crash risk) is constrained to:

$$\mathbb{E}[\ln(S_T)] = 4.60$$



**Objective:**

Formulate and solve the Maximum Entropy problem to find the continuous probability density function $p(S_T)$ for $S_T > 0$.




### 1. The Optimization Problem
We want to find the risk-neutral probability density function $p(S_T)$ for the future asset price $S_T > 0$ that maximizes Shannon's differential entropy:

$$H(p) = -\int_{0}^{\infty} p(S_T) \ln p(S_T) \, dS_T$$

Subject to our four known market-clearing constraints:
1. **Normalization:** $\int_{0}^{\infty} p(S_T) \, dS_T = 1$
2. **Risk-Free Growth (Mean Price):** $\int_{0}^{\infty} S_T p(S_T) \, dS_T = E[S_T] = 102$
3. **Market Volatility (Second Moment):** $\int_{0}^{\infty} S_T^2 p(S_T) \, dS_T = E[S_T^2] = 10,600$
4. **Asymmetric Sentiment (Log Penalty Skew):** $\int_{0}^{\infty} \ln(S_T) p(S_T) \, dS_T = E[\ln(S_T)] = 4.60$

---

### 2. The Lagrangian Formulation
We introduce four Lagrange multipliers ($\lambda_0, \lambda_1, \lambda_2, \lambda_3$) corresponding to each constraint. The functional objective $L(p)$ is established as:

$$L(p) = -\int_{0}^{\infty} p(S_T) \ln p(S_T) \, dS_T - \lambda_0 \left( \int_{0}^{\infty} p(S_T) \, dS_T - 1 \right) - \lambda_1 \left( \int_{0}^{\infty} S_T p(S_T) \, dS_T - 102 \right) - \lambda_2 \left( \int_{0}^{\infty} S_T^2 p(S_T) \, dS_T - 10,600 \right) - \lambda_3 \left( \int_{0}^{\infty} \ln(S_T) p(S_T) \, dS_T - 4.60 \right)$$

Combining the objective and constraints into a single operational integral:

$$L(p) = \int_{0}^{\infty} \left[ -p(S_T) \ln p(S_T) - \lambda_0 p(S_T) - \lambda_1 S_T p(S_T) - \lambda_2 S_T^2 p(S_T) - \lambda_3 \ln(S_T) p(S_T) \right] dS_T + \lambda_0 + 102\lambda_1 + 10600\lambda_2 + 4.60\lambda_3$$

---

### 3. Functional Differentiation
To find the global maximum, we take the functional derivative of the integrand with respect to $p(S_T)$ and set the expression to zero:

$$\frac{\partial}{\partial p} \left( -p(S_T) \ln p(S_T) - \lambda_0 p(S_T) - \lambda_1 S_T p(S_T) - \lambda_2 S_T^2 p(S_T) - \lambda_3 \ln(S_T) p(S_T) \right) = 0$$

Applying the product rule $\frac{d}{dp}[p \ln p] = \ln p + 1$:
$$-\ln p(S_T) - 1 - \lambda_0 - \lambda_1 S_T - \lambda_2 S_T^2 - \lambda_3 \ln(S_T) = 0$$

Isolating the log-density profile:
$$\ln p(S_T) = -(1 + \lambda_0) - \lambda_1 S_T - \lambda_2 S_T^2 - \lambda_3 \ln(S_T)$$

Exponentiating both sides defines the formal functional shape of our customized density distribution:
$$p(S_T) = \exp\left(-(1 + \lambda_0)\right) \cdot \exp\left(-\lambda_3 \ln(S_T)\right) \cdot \exp\left(-\lambda_1 S_T - \lambda_2 S_T^2\right)$$

Using algebraic properties ($e^{-\lambda_3 \ln(S_T)} = S_T^{-\lambda_3}$) and gathering constants under a normalization factor $A = e^{-(1+\lambda_0)}$, we achieve:

$$\boxed{p(S_T) = A \cdot S_T^{-\lambda_3} \cdot \exp\left(-\lambda_1 S_T - \lambda_2 S_T^2\right)}$$

### 4. Interpretation of the Solution Market Dynamics

The resulting Maximum Entropy probability distribution generalizes beyond rigid frameworks into a hybrid parameter system:

* **The $S_T^{-\lambda_3}$ Component:** This creates an asymmetric power-law scaling factor. When downstream put options are priced heavily, $\lambda_3$ adjusts to alter the steepness near zero, forcing a heavy, non-Gaussian **fat tail** on the left side to accurately price downward crash risks.
* **The $\exp(-\lambda_1 S_T - \lambda_2 S_T^2)$ Component:** This acts as a modified Gaussian envelope. If the asymmetric log-skew constraint were absent ($\lambda_3 = 0$), this distribution would compress neatly back into a truncated normal distribution to satisfy standard mean-variance rules.

### 5. Solving for the Numerical Values (Methodology)
Because the constraints interact non-linearly over the semi-infinite domain $S_T \in (0, \infty)$, analytical closed-form integration expressions for the multipliers do not exist. To find the exact parameters for Asset X, we minimize the dual convex objective function using a non-linear optimizer (such as `scipy.optimize` in Python):

$$\min_{\lambda_1, \lambda_2, \lambda_3} \ln \left( \int_{0}^{\infty} S_T^{-\lambda_3} \exp\left(-\lambda_1 S_T - \lambda_2 S_T^2\right) dS_T \right) + 102\lambda_1 + 10600\lambda_2 + 4.60\lambda_3$$
